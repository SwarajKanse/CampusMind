# 🔍 Statistics Experiment 12: Principal Component Analysis (PCA) & K-Means Clustering
## 2D Dimensionality Reduction of Query Embeddings

### Objective
Reduce 384-dimensional dense semantic vectors (from `all-MiniLM-L6-v2`) to 2 orthogonal principal components, and apply K-Means clustering to discover semantic topic groupings.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

# 3 synthetic clusters in 384-D vector space
np.random.seed(42)
c1 = np.random.normal(loc=0.5, scale=0.1, size=(25, 384))  # Academic / Algorithms
c2 = np.random.normal(loc=-0.5, scale=0.1, size=(25, 384)) # Admin / Timetable
c3 = np.random.normal(loc=0.0, scale=0.1, size=(25, 384))  # General greetings

X = np.vstack([c1, c2, c3])

# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Apply KMeans
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X)

print(f"Explained Variance Ratio: PC1 = {pca.explained_variance_ratio_[0]:.3f}, PC2 = {pca.explained_variance_ratio_[1]:.3f}")

# Plot
plt.figure(figsize=(8, 5))
colors = ['#7c3aed', '#06b6d4', '#22c55e']
for k in range(3):
    plt.scatter(X_pca[labels == k, 0], X_pca[labels == k, 1], c=colors[k], label=f'Cluster {k+1}', s=60, alpha=0.8)

plt.title('2D PCA Projection & Clustering of CampusMind Queries')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.grid(True, alpha=0.2)
plt.show()
```
